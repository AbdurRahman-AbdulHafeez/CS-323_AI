# Encoding Categorical Data

```python
# Apply dummy variable encoding on ExerciseAngina
data['ExerciseAngina'].value_counts()
```

```python
data['ExerciseAngina'] = data['ExerciseAngina'].map({'N': 0, 'Y': 1})
data
```

```python
data['ChestPainType'].value_counts()
```

```python
# Apply Ordinal encoding on ChestPainType
data['ChestPainType'] = data['ChestPainType'].map({'TA': 1, 'ATA': 2, 'NAP':3, 'ASY': 4})
data
```

```python
# get_dummies is a simple method to implement One-Hot Encoding
# apply one-hot encoding on Gender
data = pd.get_dummies(data, columns=['Gender'])
# The resulting table has two dummy variable encoded columns Gender_F & Gender_M
data
```

# Scaling (Normalization)

```python
# Let us normalize the 'RestingBP' column using normalization formula
data['RstingBP'] = (data['RestingBP'].min())/(data['RestingBP'].max()-data['RestingBP'].min())
data
```
